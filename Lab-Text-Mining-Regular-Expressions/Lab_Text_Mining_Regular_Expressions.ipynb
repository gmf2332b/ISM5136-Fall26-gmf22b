```python
#Step 1.1

import re
import pandas as pd

file_path = 'fraudulent_emails.txt'

data = []

from_pattern = re.compile(r'^From:\s*["\']?\s*((?i:dr\.?)\b[^<"\n]+?)\s*["\']?\s*<([a-zA-Z0-9._%+-]+@([a-zA-Z0-9.-]+\.(?:com|net)))>',re.MULTILINE,)

with open(file_path, 'r', encoding='latin-1') as f:
    for line_num, line in enumerate(f, start=1):
        match = from_pattern.match(line)
        if match:
            name = match.group(1).strip()
            email = match.group(2).strip()
            domain = match.group(3).strip()
            data.append({'Line Number': line_num,'Name': name,'E-mail Address': email,'E-mail Domain': domain,})

df = pd.DataFrame(data)

print('Total matched records:',len(df))
df.head(21)
```

```python
#Step 1.2

import re
import pandas as pd

date_pattern = re.compile(r"^Date:\s*(?:([A-Za-z]{3}),\s*)?(\d{1,2}\s+[A-Za-z]{3}\s+\d{4})\s+(\d{2}:\d{2}:\d{2})")

data = []

with open(file_path, 'r', encoding='latin-1') as f:
    lines = f.readlines()

for line_idx, line in enumerate(lines):
    from_match = from_pattern.match(line)
    if from_match:
        line_num = line_idx + 1
        name = from_match.group(1).strip()
        email = from_match.group(2).strip()
        domain = from_match.group(3).strip()

        day, date_val, time_val = None, None, None

        start_idx = max(0, line_idx - 6)
        end_idx = min(len(lines), line_idx + 7)

        for nearby_idx in range(start_idx, end_idx):
            date_match = date_pattern.match(lines[nearby_idx])
            if date_match:
                day = date_match.group(1) if date_match.group(1) else ''
                date_val = date_match.group(2)
                time_val = date_match.group(3)
                break

        data.append({'Line Number': line_num,'Name': name,'E-mail Address': email,'E-mail Domain': domain,'Day': day,'Date': date_val,'Time': time_val,})

df = pd.DataFrame(data)

df_options = df[['Name','E-mail Address','E-mail Domain','Day','Date','Time','Line Number',]]

df_options
```
