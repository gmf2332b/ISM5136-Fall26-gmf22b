```python
#Function 1
import numpy as np
import pandas as pd

def standardizer(arr):
    mean = np.mean(arr, axis=0)
    std = np.std(arr, axis=0)
    std_adjusted = np.where(std == 0, 1.0, std)
    standardized_arr = (arr - mean)/std_adjusted
    return standardized_arr
```

```python
#Function 2
def calculateDistance(obj1, obj2):
    return np.sqrt(np.sum((obj1 - obj2) ** 2))
```

```python
#Function 3
def findClosestCentroid(distArray):
    min_dist = distArray[0]
    closest_index = 0
    
    for idx in range(1, len(distArray)):
        if distArray[idx] < min_dist:
            min_dist = distArray[idx]
            closest_index = idx
    return int(np.argmin(distArray))
```

```python
#Function 4
def calculateCentroids(assignmentArr):
    return np.mean(assignmentArr, axis=0)
```

```python
#Function 5
def main(filepath='k-means_question_sample_data.csv', k=3, max_iter=100, seed=42):
    raw_data = np.genfromtxt(filepath, delimiter=",")
    print('Loaded dataset shape:',raw_data.shape)

    standardized_data = standardizer(raw_data)

    if seed is not None:
        np.random.seed(seed)
    n_samples, n_features = raw_data.shape
    initial_indices = np.random.choice(n_samples, size=k, replace=False)
    centroids = standardized_data[initial_indices].copy()
    assignments = np.zeros(n_samples, dtype=int)
    for i in range(max_iter):
        for j in range(n_samples):
            distance_to_centroids = [calculateDistance(standardized_data[j], centroids[c]) for c in range(k)]
            assignments[j] = findClosestCentroid(distance_to_centroids)
        new_centroids = np.zeros((k, n_features))
        for cluster_id in range(k):
            cluster_points = standardized_data[assignments == cluster_id]
            if len(cluster_points) > 0:
                new_centroids[cluster_id] = calculateCentroids(cluster_points)
            else:
                new_centroids[cluster_id] = centroids[cluster_id]

        if np.allclose(centroids, new_centroids, atol=1e-6):
            print('Convergence achieved at iteration', i + 1)
            centroids = new_centroids
            break

        centroids = new_centroids

    counts = pd.Series(assignments).value_counts().sort_index()
    print('\nCluster assignment counts - ')
    for cluster_id, count in counts.items():
        print('Cluster', cluster_id,':' ,count, 'points')
    return centroids, assignments


if __name__ == "__main__":
    final_centroids, final_assignments = main()
```
