```python
import pandas as pd
#Step 1
#1.1
df = pd.read_table('RiskComplete.txt')
```

```python
#Step 2
#2.1
df = df.drop_duplicates()
```

```python
#2.2
df = df.drop(['ID'], axis = 1)
df
```

```python
#2.3
missing = df.isnull().sum()
```

```python
#Question 1
print('Missing:\n', missing)
df['INCOME'] = df['INCOME'].fillna(round(df['INCOME'].mean()),)
df['GENDER'] = df['GENDER'].fillna(df['GENDER'].mode()[0])
df['NUMCARDS'] = df['NUMCARDS'].fillna(round(df['NUMCARDS'].mean()))
df['LOANS'] = df['LOANS'].fillna(round(df['LOANS'].mean()))
print('\nMissing values replaced:\n',df.isnull().sum())
```

```python
#Question 2
print('Loan Average:',round((df['LOANS']).mean(),4))
```

```python
#2.4
import numpy as np
df = df.replace('unknown', np.nan)
df = df.dropna()
```

```python
#Question 3
print(df.shape[0])
```

```python
#Step 3
#Question 4
print('Ratio of females:',round((df['GENDER'] == 'f').mean(),4))
```

```python
#Question 5
print(df['VALUE'].value_counts())
```

```python
#Step 4
#4.1
df['VALUE'] = df['VALUE'].replace({'strong profit' : 'profit', 'weak profit':'profit', 'bad loss': 'loss'})
```

```python
#Question 6
print(df['VALUE'].value_counts())
```

```python
#4.2
inputs = ['GENDER', 'MARITAL', 'HOWPAID', 'MORTGAGE']
df = pd.get_dummies(df, columns=inputs, drop_first = True, dtype = int)
df['VALUE'] = df['VALUE'].replace({'profit':1, 'loss':0})
print(df['VALUE'].value_counts())
```

```python
#4.3
from sklearn.preprocessing import StandardScaler, MinMaxScaler
stscaler = StandardScaler()
standardized = ['AGE', 'INCOME']
df[standardized] = stscaler.fit_transform(df[standardized])
mmScaler = MinMaxScaler()
normalized = ['NUMKIDS','NUMCARDS','STORECAR','LOANS']
df[normalized] = mmScaler.fit_transform(df[normalized])
```

```python
#Question 7
print('Standardized INCOME mean:',round(df['INCOME'].mean(),4))
print('Normalized NUMCARDS mean:',round(df['NUMCARDS'].mean(),4))
```

```python
#Step 5
#5.1
x = df.drop('VALUE', axis = 1)
y = df['VALUE']

display(x.head(10))
display(y.head(10))
```

```python
#5.2
from sklearn.neural_network import MLPClassifier
features = x.shape[1]
modelNN = MLPClassifier(hidden_layer_sizes=(features,features,features),activation='logistic',solver='sgd',learning_rate_init=0.3,random_state = 1, shuffle=False)
```

```python
#5.3
import time
start_time = time.time()
modelNN.fit(x,y)
end_time = time.time()
```

```python
#Question 9
print('Model trained in', round(end_time - start_time, 2), 'seconds')
```

```python
#Question 10
print('Total layers in model:', modelNN.n_layers_)
print('Number of output nodes:', modelNN.n_outputs_)
weight = modelNN.coefs_[0][9][0]
print('The weight between the 10th input and the 1st neuron of the first hidden layer is',weight)
```

```python
#Question 11
modelNN2 = MLPClassifier(hidden_layer_sizes=(500,500,500),activation='logistic',solver='sgd',learning_rate_init=0.3,random_state = 1, shuffle=False)
start_time = time.time()
modelNN2.fit(x,y)
end_time = time.time()
print('Model trained in', round(end_time - start_time, 2), 'seconds')
# Since the model went from 10 to 500 neurons per layer, it took much longer to train. Even with putting 500 neurons over three layers 
# on a dataset of around 4,100 rows, this neural network has more parameters than necessary.
# This could lead to overfitting within the model where it is now memorizing noise in the training set and is 
# less likely to generalize patterns.
```
